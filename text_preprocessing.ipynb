```python
# Task 2: Text Preprocessing
# Original input text

text = """What about this? Google is making complex and multi-step tasks more accessible with its latest AI model, Gemini 4 Neon. The model 
announced on Wednesday, November 18, combines frontier-level intelligence with fast end-to-end task completions across complex software 
engineering, knowledge work, and science workflows. Neon is offering advanced reasoning at better speeds and lower costs.

The model brings improvements across use cases such as coding, data analysis, research, and more. According to Google, Gemini 4 Neon ranked 
ahead of its peers and is at number three with a score of 61 on the Artificial Intelligence Index, an independent benchmarking platform. It is 
aimed at users who need powerful models to perform demanding tasks but don’t necessarily need the full capabilities of Google’s top-of-the-line 
AI models. Along with the launch, Google has also replaced Gemini 4 Argon in the Gemini app and shuttered the Gemini 3.1 Pro model. 

Gemini 4 Neon: Performance
The model ranked second on the Vals Index and topped benchmarks for legal research and tax-related tasks. On the Agent Arena Index, the 
model secured the second rank for its ability to adapt whenever a user changes or corrects instructions when a task is in progress. Google 
claims Neon is also good at handling tasks that require multi-steps and has made fewer mistakes in its latest internal tests, recording its 
lowest rate of making up information so far. The tech giant said that Gemini 4 Neon operates about three times faster than Gemini 4 Carbon while 
reducing overall costs by around 60 per cent. The model has been developed to manage complex workplace tasks that involve an assortment of 
documents across formats. It is capable of analysing PDFs, charts, audio, videos, and can analyse all these inputs together and turn them into 
useful outputs like financial models, reports, or research documents. It can process up to 1 million tokens of information at a time. This makes 
it particularly beneficial for enterprises, as per the company."""

print(text)
```

```python
# Tokenization: Split the text into individual words or sentences.
from nltk.tokenize import word_tokenize

# Tokenize the input text into individual words
tokens = word_tokenize(text)

print("Tokens:", tokens)
```

```python
# Lowercasing: Convert all text to lowercase.
lowercased = [word.lower() for word in tokens]

print("Lowercased:", lowercased)
```

```python
# Stopword Removal: Remove common words that do not carry significant meaning.
from nltk.corpus import stopwords

# Get the English stopwords
stop_words = set(stopwords.words('english'))

# Remove stopwords from the lowercased tokens
filtered_words = [
    word for word in lowercased
    if word not in stop_words
]

print("Without Stopwords:", filtered_words)
```

```python
# Punctuation Removal: Remove punctuation marks from the filtered words.
import string

# Get all standard punctuation marks
punctuation = string.punctuation

# Remove punctuation from the filtered words
cleaned_words = [
    word.translate(str.maketrans('', '', string.punctuation))
    for word in filtered_words
]

cleaned_words = [
    word for word in cleaned_words if word
]

print("After Punctuation Removal:", cleaned_words)
```

```python
# Stemming: Reduce words to their root form.
from nltk.stem import PorterStemmer

# Initialize Porter Stemmer
stemmer = PorterStemmer()

# Perform stemming on cleaned words
stemmed_words = [
    stemmer.stem(word) for word in cleaned_words
]

print("Stemmed Words:", stemmed_words)
```

```python
# Lemmatization: Reduce words to their base or dictionary form.
import spacy

# Load the English language model
nlp = spacy.load("en_core_web_sm")

# Convert cleaned words into a spaCy document
doc = nlp(" ".join(cleaned_words))

# Perform lemmatization
lemmatized_words = [
    token.lemma_ for token in doc
]

print("Lemmatized Words:", lemmatized_words)
```

```python
# Task 3: Analyzing Word Frequency
text = """
Artificial Intelligence has transformed the way researchers analyze large volumes of unstructured data. Modern Natural Language Processing
systems are designed to understand human languages, recognize patterns,and generate meaningful responses.
Researchers have been developing sophisticated machine learning algorithms that process millions of documents, identify relationships
between words, and extract valuable information. These technologies are increasingly being used in healthcare, education, finance, and
scientific research.
Although intelligent systems can perform complex linguistic tasks, they still encounter challenges when interpreting ambiguous sentences,
context-dependent expressions, and rapidly evolving vocabulary. Therefore, effective text preprocessing techniques, including
tokenization, stopword removal, stemming, and lemmatization, play an essential role in improving the quality of language-based models."""

print(text)
```

```python
import re
import string
from collections import Counter
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords

# Step 1: Convert text to lowercase
lowercase_text = text.lower()

# Step 2: Tokenize the new text
tokens = word_tokenize(lowercase_text)

# Step 3: Remove punctuation
words = [
    word for word in tokens
    if word not in string.punctuation
]

# Step 4: Remove stopwords
stop_words = set(stopwords.words('english'))

cleaned_words = [
    word for word in words
    if word not in stop_words
]

# Step 5: Calculate word frequencies
word_frequency = Counter(cleaned_words)

# Step 6: Display the top 5 most frequent words
top_words = word_frequency.most_common(5)

print("Top 5 Word Frequencies:")
for word, frequency in top_words:
    print(f"{word}: {frequency}")
```
